# Exercise 10: Knowledge Graph in Kanren

Write kanren goals to query the knowledge graph developed in the lectures to:

* List all students in each of the departments.
* Find all staff that teach student "Alan Turing"
* Find all students taught by staff member "Prof Hui Wang".
* Find all academic staff involved in delivery of a department's programmes.

```python
import csv
from kanren import facts, Relation
d = dict()
with open('Simple Graph Triples.csv') as csvfile:
    triples = csv.reader(csvfile)
    # Put the triples in a dictionary keyed by the relations
    for triple in triples:
        if triple[1] not in d.keys():
            d[triple[1]] = [(triple[0],triple[2])]
        else:
            d[triple[1]].append((triple[0],triple[2]))


# Now we create the facts. This is a little tricky.
# We first add the relations to the dictionary of global variables.
# We then pass the list of facts in using * to unpack the list into its members
for k in d.keys():
    globals()[k] = Relation()
    facts(globals()[k],*d[k])
```

* List all students in each of the departments.

```python
from kanren import vars, run
x, y = vars(2)
goals = (
    is_enrolled_in(x,y),
    offers_programme("EEECS",y)
)
run(0, x, goals)
```

* Find all staff that teach student "Alan Turing"

```python
from kanren import var
z = var()
goals = (
    is_enrolled_in("Alan Turing",y),
    has_module(y,z),
    is_taught_by(z,x)
)
run(0, x, goals)
```

* Find all students taught by staff member "Prof Hui Wang".

```python
goals = (
    is_taught_by(z,"Prof Hui Wang"),
    has_module(y,z),
    is_enrolled_in(x,y),
)
run(0, x, goals)
```

* Find all academic staff involved in delivery of a department's programmes.

```python
from kanren import conde

goals = conde(
    (offers_programme("EEECS",y), has_module(y,z), is_taught_by(z,x)),
    (offers_programme("EEECS",y), is_directed_by(y,x))
)
run(0, x, goals)
```
