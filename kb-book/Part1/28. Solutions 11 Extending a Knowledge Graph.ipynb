1. Extend the ontology and knowledge base developed in the lectures to include a new relation `employed_by({staff},{school})`. [Hint: you will need to use the prefix `onto` when specifying the domain and range of the relation].

```python
with onto:
    class employed_by(ObjectProperty):
        domain = [onto.Staff]
        range = [onto.School]
```

2. Populate the graph with the following facts

* Employed by EEECS: Dr Barry Devereux, Dr Ihsen Alouani, Dr Lu Bai, Prof Hui Wang, Prof Iain Styles, Dr Yang Hua, Dr Reza Rafiee, Dr Niall McLaughlin, Dr Ciara Rafferty, Dr Ayesha Khalid, Dr Amy Liu, Prof Maire O'Neill, Dr Neil Anderson, Dr Chao Tian,

* Employed by Math & Physics: Dr Lisa McFetridge, Dr Hannah Mitchell

```python
employees = {
    'Dr Barry Devereux': 'EEECS',
    'Dr Ihsen Alouani': 'EEECS',
    'Dr Lu Bai': 'EEECS',
    'Prof Hui Wang': 'EEECS',
    'Prof Iain Styles': 'EEECS',
    'Dr Yang Hua': 'EEECS',
    'Dr Reza Rafiee': 'EEECS',
    'Dr Niall McLaughlin': 'EEECS',
    'Dr Ciara Rafferty': 'EEECS',
    'Dr Ayesha Khalid': 'EEECS',
    'Dr Amy Liu': 'EEECS',
    "Prof Maire O'Neill": 'EEECS',
    'Dr Neil Anderson': 'EEECS',
    'Dr Chao Tian': 'EEECS',
    "Dr Lisa McFetridge": 'Maths and Physics',
    "Dr Hannah Mitchell": 'Maths and Physics'
}

# Get all the staff objects
stafflist = list(default_world.sparql(
    """
    PREFIX RDF: <http://www.dummy.info/new.owl#>
        SELECT ?staff
        WHERE{
            ?staff rdf:type RDF:Staff
        }
        """))
print(stafflist)

schoollist = list(default_world.sparql(
    """
    PREFIX RDF: <http://www.dummy.info/new.owl#>
        SELECT ?school
        {
            ?school rdf:type RDF:School
        }
        """))
print(schoollist)

schools = dict()
for s in schoollist:
    schools[s[0].school_name[0]] = s

staff = dict()
for s in stafflist:
    staff[s[0].person_name[0]] = s

print(schools)
with onto:
    for employee in employees.keys():
        print(employee)
        school = employees[employee]
        print(school)
        print(staff[employee][0])
        print(schools[school][0])
        getattr(staff[employee][0],'employed_by').append(schools[school][0])
        print(getattr(staff[employee][0],'employed_by'))
```

3. Write SPARQL queries to find out the following:

* All staff employed by EEECS teaching on programs owned by Maths and Physics

```python
x = list(default_world.sparql(
    """
    PREFIX ONTO: <http://www.dummy.info/new.owl#>
        SELECT ?staff ?module ?program ?school
        {
            ?staff rdf:type ONTO:Staff
            ?module rdf:type ONTO:Module
            ?program rdf:type ONTO:Program
            ?school rdf:type ONTO:School
            ?employer rdf:type ONTO:School
            ?module ONTO:is_taught_by ?staff
            ?program ONTO:has_module ?module
            ?school ONTO:offers_programme ?program
            ?staff rdf:type ONTO:Staff
            ?employer ONTO:school_name "EEECS"
            ?school ONTO:school_name "Maths and Physics"
            ?staff ONTO:employed_by ?employer
        }
    """))
for i in x:
    print(f"{i[0].person_name [0]} teaches module {i[1].module_title[0]} on program {i[2].program_title[0]} offered by school {i[3].school_name[0]}")
```
