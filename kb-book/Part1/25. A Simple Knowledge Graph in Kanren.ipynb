# Developing a simple Knowledge Graph in Kanren


Consider the following familiar situation. A University department teaches students, who are enrolled on different modules.

* A University has multiple departments
* A department offers a range of programmes
* Each programme is composed of modules
* Students enroll on a single programme.
* Each programme has an academic director.
* Each module has a teacher.
* Students take multiple modules

We therefore have the following entities:

* `department`
* `programme`
* `module`
* `student`
* `academic`

and predicates

* `offers_programme(department, programme)`
* `has_module(programme, module)`
* `is_enrolled_on(student,programme)`
* `is_taught_by(module,academic)`
* `is_directed_by(programme,academic)`

and ontology (in graphical form)

![Simple ontology of academic programme structure](../images/academic_graph.png)


Kanren provides us with mechanisms for directly representing these, but these mechanisms are limited. We can specify the type of relation/predicate, and we can create instances of those relation, but there is no mechanism for specifying the ontology.

```python
from kanren import facts, Relation, run, var
x = var()
offers_programme = Relation()
has_module = Relation()
is_directed_by = Relation()
is_taught_by = Relation()
is_enrolled_in = Relation()

f = [("Math & Physics", "MSc Data Analytics and AI"),
      ("EEECS", "MSc Artificial Intelligence"),
      ("EEECS", "MSc Cyber Security")]

facts(offers_programme, *f)

run(0,x,offers_programme("EEECS",x))
```

This will very quickly become tedious, so we will instead load the data from a pre-prepared CSV file

```python
import csv
from kanren import facts, Relation
d = dict()
with open('Simple Graph Triples.csv') as csvfile:
    triples = csv.reader(csvfile)
    # Put the triples in a dictionary keyed by the relations
    for triple in triples:
        if triple[1] not in d.keys():
            d[triple[1]] = [(triple[0],triple[2])]
        else:
            d[triple[1]].append((triple[0],triple[2]))


# Now we create the facts. This is a little tricky.
# We first add the relations to the dictionary of global variables.
# We then pass the list of facts in using * to unpack the list into its members
for k in d.keys():
    globals()[k] = Relation()
    facts(globals()[k],*d[k])
```

## Querying the graph

Now we have the graph, we can write some queries to extract information from it. Let's start with a simple query - find all the students on a programme.
