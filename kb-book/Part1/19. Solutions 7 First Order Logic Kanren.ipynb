1. Use Kanren to solve the following logic puzzle:

Ahmed, Chen, and Niamh are three friends. They went for dinner together one evening. They decided that they wanted to try all of the items on the menu so they all chose a different main course and a different dessert. The menu options for the main course were Pizza, Daal, and Falafel. For dessert, the choices were Apple Pie, Cheesecake, and Ice Cream. Given the five statements below, can you work out who ordered what?

1. The person who had the Pizza did not have the Apple Pie.
1. Niamh had the Daal.
1. Ahmed did not have Cheesecake.
1. Chen had the Apple Pie.
1. The person who had the Ice Cream did not have the Falafel.

```python
from kanren import run, var, membero, eq, lall, lany

diners = var()

rules = lall(
    eq(
        (
            ("Ahmed",var(),var()),
            ("Chen",var(),var()),
            ("Niamh",var(),var())
        ),
        diners
    ),
    lany(
        membero((var(),"Pizza","Cheesecake"),diners),
        membero((var(),"Pizza","Ice Cream"),diners)
    ),
    membero(("Niamh","Daal",var()),diners),
    lany(
        membero(("Ahmed",var(),"Apple Pie"),diners),
        membero(("Ahmed",var(),"Ice Cream"),diners)
    ),
    membero(("Chen",var(),"Apple Pie"),diners),
    lany(
        membero((var(),"Pizza","Ice Cream"),diners),
        membero((var(),"Daal","Ice Cream"),diners)
    ),
    membero((var(),"Falafel",var()),diners),
    membero((var(),var(),"Cheesecake"),diners),
)

solutions = run(0, diners, rules)
print(solutions)
```

2. Einstein's Riddle

Write a Kanren program to solve Einsteins riddle. There are five houses in a row. Each house is a different colour and is owned by a person of a different nationality. Each person keeps a different animal as a pet, prefers a different drink, and smokes a different brand of cigarette. Fifteen facts are known about the people and their houses:

1. There are five houses.
1. The Englishman lives in the red house.
1. The Spaniard owns the dog.
1. Coffee is drunk in the green house.
1. The Ukrainian drinks tea.
1. The green house is immediately to the right of the ivory house.
1. The Old Gold smoker owns snails.
1. Kools are smoked in the yellow house.
1. Milk is drunk in the middle house.
1. The Norwegian lives in the first house.
1. The man who smokes Chesterfields lives in the house next to the man with the fox.
1. Kools are smoked in the house next to the house where the horse is kept.
1. The Lucky Strike smoker drinks orange juice.
1. The Japanese smokes Parliaments.
1. The Norwegian lives next to the blue house.
 
Who drinks water? Who owns the zebra?


```python
from kanren import var, vars, lall, lany, membero, eq, run, conde

houses = var()

rules = lall(
    eq(
        (
            (1,var(),var(),var(),var(),var()),
            (2,var(),var(),var(),var(),var()),
            (3,var(),var(),var(),var(),var()),
            (4,var(),var(),var(),var(),var()),
            (5,var(),var(),var(),var(),var()),
        ), houses
    ),
    membero((var(),"Englishman", var(), var(), var(), "Red"), houses),
    membero((var(),"Spaniard", "Dog", var(), var(), var()), houses),
    membero((var(),var(), var(), "Coffee", var(), "Green"), houses),
    membero((var(),"Ukrainian", var(), "Tea", var(), var()), houses),
    conde(
        (membero((1,var(),var(),var(),var(),"Ivory"),houses),membero((2,var(),var(),var(),var(),"Green"),houses)),
        (membero((2,var(),var(),var(),var(),"Ivory"),houses),membero((3,var(),var(),var(),var(),"Green"),houses)),
        (membero((3,var(),var(),var(),var(),"Ivory"),houses),membero((4,var(),var(),var(),var(),"Green"),houses)),
        (membero((4,var(),var(),var(),var(),"Ivory"),houses),membero((5,var(),var(),var(),var(),"Green"),houses))
    ),
    membero((var(),var(), "Snails", var(), "Old Gold", var()), houses),
    membero((var(),var(), var(), var(), "Kools", "Yellow"), houses),
    membero((3,var(),var(),"Milk",var(),var()), houses),
    membero((1,"Norwegian",var(),var(),var(),var()), houses),
    conde(
        (membero((1,var(),var(),var(),"Chesterfields",var()),houses), membero((2,var(),"Fox",var(),var(),var()),houses)),
        (membero((2,var(),var(),var(),"Chesterfields",var()),houses), membero((1,var(),"Fox",var(),var(),var()),houses)),
        (membero((2,var(),var(),var(),"Chesterfields",var()),houses), membero((3,var(),"Fox",var(),var(),var()),houses)),
        (membero((3,var(),var(),var(),"Chesterfields",var()),houses), membero((2,var(),"Fox",var(),var(),var()),houses)),
        (membero((3,var(),var(),var(),"Chesterfields",var()),houses), membero((4,var(),"Fox",var(),var(),var()),houses)),
        (membero((4,var(),var(),var(),"Chesterfields",var()),houses), membero((3,var(),"Fox",var(),var(),var()),houses)),
        (membero((4,var(),var(),var(),"Chesterfields",var()),houses), membero((5,var(),"Fox",var(),var(),var()),houses)),
        (membero((5,var(),var(),var(),"Chesterfields",var()),houses), membero((4,var(),"Fox",var(),var(),var()),houses))
    ),
    conde(
        (membero((1,var(),var(),var(),"Kools",var()),houses), membero((2,var(),"Horse",var(),var(),var()),houses)),
        (membero((2,var(),var(),var(),"Kools",var()),houses), membero((1,var(),"Horse",var(),var(),var()),houses)),
        (membero((2,var(),var(),var(),"Kools",var()),houses), membero((3,var(),"Horse",var(),var(),var()),houses)),
        (membero((3,var(),var(),var(),"Kools",var()),houses), membero((2,var(),"Horse",var(),var(),var()),houses)),
        (membero((3,var(),var(),var(),"Kools",var()),houses), membero((4,var(),"Horse",var(),var(),var()),houses)),
        (membero((4,var(),var(),var(),"Kools",var()),houses), membero((3,var(),"Horse",var(),var(),var()),houses)),
        (membero((4,var(),var(),var(),"Kools",var()),houses), membero((5,var(),"Horse",var(),var(),var()),houses)),
        (membero((5,var(),var(),var(),"Kools",var()),houses), membero((4,var(),"Horse",var(),var(),var()),houses))
    ),
    membero((var(),var(), var(), var(), "Kools", "Yellow"), houses),
    membero((var(),"Japanese", var(), var(), "Parliaments", var()), houses),
    membero((var(),var(), var(), "Orange Juice", "Lucky Strike", var()), houses),
    conde(
        (membero((1,"Norwegian",var(),var(),"Kools",var()),houses), membero((2,var(),var(),var(),var(),"Blue"),houses)),
        (membero((2,"Norwegian",var(),var(),"Kools",var()),houses), membero((1,var(),var(),var(),var(),"Blue"),houses)),
        (membero((2,"Norwegian",var(),var(),"Kools",var()),houses), membero((3,var(),var(),var(),var(),"Blue"),houses)),        
        (membero((3,"Norwegian",var(),var(),"Kools",var()),houses), membero((2,var(),var(),var(),var(),"Blue"),houses)),
        (membero((3,"Norwegian",var(),var(),"Kools",var()),houses), membero((4,var(),var(),var(),var(),"Blue"),houses)),
        (membero((4,"Norwegian",var(),var(),"Kools",var()),houses), membero((3,var(),var(),var(),var(),"Blue"),houses)),
        (membero((4,"Norwegian",var(),var(),"Kools",var()),houses), membero((5,var(),var(),var(),var(),"Blue"),houses)),
        (membero((1,"Norwegian",var(),var(),"Kools",var()),houses), membero((4,var(),var(),var(),var(),"Blue"),houses)),
    ),
    membero((var(),var(), var(), "Water", var(), var()), houses),
    membero((var(),var(), "Zebra", var(), var(), var()), houses)
)

solutions = run(0, houses, rules)
for i in solutions[0]:
    print(i)

```
