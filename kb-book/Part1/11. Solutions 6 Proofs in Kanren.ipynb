# Exercises 5: Verifying Proofs in Kanren


1. Write a kanren program to verify the argument $P\implies Q\land R,\ R\implies S,\ P,\ Q\ :\ S$

```python
from kanren import vars, run, eq, lall, lany
P,Q,R,S = vars(4)
goals = lall(
    lany(eq(P,False),lall(eq(Q,True),eq(R,True))),
    lany(eq(R,False),eq(S,True)),
    eq(P, True),
    eq(Q, True)
)

run(0, S, goals)
```

2. Write a kanren program to verify the argument $\lnot P\implies \lnot Q,\ Q,\ P\implies S\land Q\ :\ S $

```python
goals = lall(
    lany(eq(P,True),eq(Q,False)),
    eq(Q, True),
    lany(eq(P,False),lall(eq(S,True),eq(Q,True)))
)
run(0, S, goals)
```

3. Write a kanren program to verify that the following argument is invalid: $\lnot(P\land Q)\implies S,\ \lnot S,\ P\implies\lnot Q\ :\ \lnot Q$

```python
goals = lall(
    lany(lall(eq(P,True), eq(Q,True)), eq(S,True)),
    eq(S, False),
    lany(eq(P, False), eq(Q, False))
)
run(0, Q, goals)
```


4. Write a kanren program to verify the argument $(P\land Q)\implies R,\ \lnot R,\ P\ :\ \lnot Q$

```python
goals = lall(
    lany(lany(eq(P,False),eq(Q,False)),eq(R,True)),
    eq(R, False),
    eq(P, True)       
)

run(0, Q, goals)
```

5. Write a kanren program to verify that the following argument is invalid: $P\iff Q,\ P,\ Q\implies\lnot R,\ R\ :\ Q$

```python
goals=(
    lany(lall(eq(P,True),eq(Q,True)), lall(eq(P,False),eq(Q,False))),
    eq(P,True),
    lany(eq(Q,False),eq(R,False)),
    eq(R,True)
)
run(0, Q, goals)
```

6. Write a kanren program to verify the argument $P\implies Q,\ \lnot Q\ :\ P\implies R$. You may find the following equivalence helpful: $A\implies B \equiv \lnot A \lor B$.

```python
# Have to do this in two parts
goals=(
    lany(eq(P,False),eq(Q,True)),
    eq(Q,False),
    lany(eq(P,False),eq(R,True))
)
run(0, [P,Q,R], goals)
```

7. Write a kanren program to verify the argument $P\land Q,\ Q\implies\lnot R\ :\ \lnot(R\land\lnot S)$

```python
goals=(
    lall(eq(P,True),eq(Q,True)),
    lany(eq(Q,False),eq(R,False)),
    lany(eq(R,False),eq(S,True))
)
run(0,[P,Q,R,S],goals)
```

8. Write a kanren program to verify the argument $\lnot P,\ \lnot Q\ :\ P\iff Q$

```python
goals=(
    eq(P,False),
    eq(Q,False),
    lany(lall(eq(P,True),eq(Q,True)),lall(eq(P,False),eq(Q,False)))
)
run(0,[P,Q,R,S],goals)
```

9. Write a kanren program to verify the argument $P\implies \lnot Q \land\lnot R,\ P,\ :\ \lnot Q$

```python
goals = (
    lany(eq(P,False),lall(eq(Q,False),eq(R,False))),
    eq(P,True),
    eq(Q,False)
)
run(0,[P,Q,R],goals)
```
