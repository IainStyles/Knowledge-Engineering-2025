# Developing a simple Knowledge Graph in Kanren


Consider the following familiar situation. A University department teaches students, who are enrolled on different modules.

* A University has multiple departments
* A department offers a range of programmes
* Each programme is composed of modules
* Students enroll on a single programme.
* Each programme has an academic director.
* Each module has a teacher.
* Students take multiple modules

We therefore have the following entities:

* `department`
* `programme`
* `module`
* `student`
* `academic`

and predicates

* `offers_programme(department, programme)`
* `has_module(programme, module)`
* `is_enrolled_on(student,programme)`
* `is_taught_by(module,academic)`
* `is_directed_by(programme,academic)`

and ontology (in graphical form)

![Simple ontology of academic programme structure](../images/academic_graph.png)
