# Simple graph visualisation script

Simple python function to visualise a graph with the `d3graph` package. Documentation for this can be found [here](https://erdogant.github.io/d3graph/pages/html/index.html).

```python
!pip install d3graph
```

```python
from d3graph import d3graph, vec2adjmat

def GraphVis(source,target,fname):
    d3 = d3graph()
    adjmat = vec2adjmat(source, target)
    d3.graph(adjmat)
    d3.set_edge_properties(directed=True, marker_end='arrow', marker_start='')

    d3.show(filepath='fname')
    return None
```

```python
# Example data format
source = ['Toxin', 'Toxin', 'Drug', 'Drug', 'Gene', 'Gene', 'Gene', 'Protein','RNA','Mutation']
target = ['Disease','Symptom','Disease','Symptom','Disease','Protein','Mutation','Disease','Mutation','Protein']

GraphVis(source,target,'./vis-ex.html')


```
