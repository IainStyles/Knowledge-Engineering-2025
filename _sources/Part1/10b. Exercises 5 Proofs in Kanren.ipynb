0. Write goal constructors for material implication and bidirectional implication.

```python
from kanren import eq, lany, lall

def implies(A,B):
    # A=>B == not(A) v B
    return lany(
        eq(A, False),
        eq(B, True)
        )

def iff(A,B):
    # A<=>B == (A^B) v (not(A)^not(B))
    return lany(
        lall(
            eq(A,True),
            eq(B,True)
            ),
        lall(
            eq(A,False),
            eq(B,False)
            )
        )

# also add AND and OR for convenience

def land(A,B):
    return lall(
        eq(A,True),
        eq(B,True)
        )

def lor(A,B):
    return lany(
        eq(A,True),
        eq(B,True)
        )
```

1. Write a kanren program to verify the argument $P\implies Q\land R,\ R\implies S,\ P,\ Q\ :\ S$

```python
from kanren import vars, run
P,Q,R,S = vars(4)
goals = lall(
    implies(P,Q),
    implies(P,R),
    implies(R,S),
    eq(P, True),
    eq(Q, True)
)

run(0, S, goals)
```

2. Write a kanren program to verify the argument $\lnot P\implies \lnot Q,\ Q,\ P\implies S\land Q\ :\ S $

```python
goals = lall(
    lany(eq(P,True),eq(Q,False)),
    eq(Q, True),
    implies(P,S),
    implies(P,Q)
)
run(0, S, goals)
```

3. Write a kanren program to verify that the following argument is invalid: $\lnot(P\land Q)\implies S,\ \lnot S,\ P\implies\lnot Q\ :\ \lnot Q$

```python
goals = lall(
    lany(
        eq(
            lall(eq(P,True),eq(Q,True)),
            False),
        eq(S,True)
        ),
    eq(S, False),
    lany(eq(P,False),eq(Q,False))
)
run(0, Q, goals)
```


4. Write a kanren program to verify the argument $(P\land Q)\implies R,\ \lnot R,\ P\ :\ \lnot Q$

```python
goals = lall(
    lany(
        eq(
            lall(
                eq(P,True),
                eq(Q,True)
                ),
            False
            ),
        eq(R, True)
    ),
    eq(R, False),
    eq(P, True)       
)

run(0, Q, goals)
```
